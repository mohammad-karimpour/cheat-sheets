```python
# Estimator: یک آبجکت از (اس کی لرن) است که می‌تواند چیزی را از داده یاد بگیرد.
# یک نکته خیلی مهم: فیت همیشه به معنی یاد گرفتن مدل نیست. بلکه تعریف دقیقتر این است: اطلاعات لازم را از داده دریافت کن و object را برای استفاده آماده کن.
```

```python
# Preprocessing: تبدیل داده خام به چیزی که مناسب عملیات یادگیری باشد(اصلاح دیتاست)

# ابزار های اصلاح دیتا ست:
# StandardScaler: استاندارد سازی داده
# MinMaxScaler: کار اصلی آن این است که مقادیر را به یک بازه مشخص، معمولاً بین 0 و 1، منتقل کند.
# RobustScaler: برای مقادیر پرت استفاده میشه و انحراف معیار آنهارو کم میکنه
# SimpleImputer: پر کردن مقادیر ناموجود و گمشده با مقادیر بهینه
# OneHotEncoder: تبدیل حروف به مقادیر درست برای مدل هایی که نیمتوانند با حروف کار کنند
# LabelEncoder: مانند بالایی ولی برای عنوان ها استفاده میشه
```

```python
# مثال کلی تا اینجا

from sklearn.datasets import load_iris # دیتا ست مربوط به 150 گل
from sklearn.model_selection import train_test_split # تقسیم داده برای یادگیری و آزمایش
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier


iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
S = model.score(X_test,y_test)

pred = model.predict(X_test)

print("دیتاست:", iris.data)
print("جواب ها:", y)
print("واقعی:   ", y_test)
print("پیش‌بینی:", pred)
print(S)
```
